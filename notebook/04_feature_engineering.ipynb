```python

import pandas as pd

# Membaca dataset hasil Data Cleaning
df = pd.read_csv("../data/processed/01_clean_smartphones.csv")

# Kurs konversi INR ke IDR
kurs_inr_idr = 190.22

# Mengubah harga INR menjadi Rupiah
df["price_idr"] = (df["price_inr"] * kurs_inr_idr).round(0).astype("int64")

# Daftar kriteria untuk K-Means dan TOPSIS
features = [
    "price_idr",
    "ram_gb",
    "storage_gb",
    "clock_speed_ghz",
    "battery_mah",
    "rear_camera_main_mp",
    "charging_watt"
]

# Menyiapkan data smartphone dan kriteria
model_data = df[["smartphone_brand", "model"] + features].copy()

# Memeriksa hasil
print("Jumlah smartphone:", len(model_data))
print("Jumlah kriteria:", len(features))
print("Kriteria:", features)

print("\nJumlah nilai kosong:")
print(model_data[features].isnull().sum())

display(model_data.head())
```

```python

output_path = "../data/processed/02_feature_engineering.csv"

model_data.to_csv(output_path, index=False)

print("Data Feature Engineering berhasil disimpan!")
print("Lokasi:", output_path)
print("Jumlah data:", len(model_data))
print("Jumlah kriteria:", len(features))
```
