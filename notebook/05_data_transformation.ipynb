```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("../data/processed/02_feature_engineering.csv")

print("Ukuran dataset:", df.shape)
print("\nInformasi dataset:")
df.info()

display(df.head())
```

```python

print("Jumlah missing values:")
print(df.isnull().sum())

print("\nJumlah baris duplikat:", df.duplicated().sum())

print("\nStatistik deskriptif:")
display(df.describe())
```

```python

features = [
    "price_idr",
    "ram_gb",
    "storage_gb",
    "clock_speed_ghz",
    "battery_mah",
    "rear_camera_main_mp",
    "charging_watt"
]

X = df[features].copy()

print("Fitur yang digunakan:")
for feature in features:
    print("-", feature)

print("\nUkuran matriks fitur:", X.shape)
print("\nTipe data:")
print(X.dtypes)
```

```python

print("Statistik sebelum standardisasi:")
display(X.describe().T)

print("\nVarians setiap fitur:")
display(X.var().to_frame("variance"))
```

```python

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

df_scaled = pd.DataFrame(
    X_scaled,
    columns=features,
    index=df.index
)

print("Ukuran sebelum transformasi:", X.shape)
print("Ukuran setelah transformasi:", df_scaled.shape)

display(df_scaled.head())
```

```python

# Menyimpan fitur yang sudah distandardisasi
output_path = "../data/processed/03_data_transformation.csv"

df_scaled.to_csv(output_path, index=False)

print("Data Transformation berhasil disimpan!")
print("Lokasi:", output_path)
print("Jumlah baris:", len(df_scaled))
print("Jumlah kolom:", len(df_scaled.columns))
```
