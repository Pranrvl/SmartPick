# Business Understanding

## 1. Latar Belakang

Pemilihan smartphone dapat menjadi sulit karena terdapat banyak pilihan
dengan spesifikasi, harga, dan karakteristik yang berbeda. Pengguna
sering kali harus membandingkan banyak smartphone berdasarkan beberapa
kriteria seperti harga, RAM, penyimpanan, performa, kamera, baterai,
dan charging.

Oleh karena itu, diperlukan sebuah Sistem Pendukung Keputusan (SPK)
yang dapat membantu pengguna memilih smartphone berdasarkan kebutuhan
dan preferensi pengguna.

## 2. Permasalahan

Bagaimana membangun sistem yang dapat mengelompokkan smartphone
berdasarkan karakteristiknya dan memberikan rekomendasi smartphone
berdasarkan preferensi pengguna?

## 3. Tujuan

Membangun sistem pendukung keputusan pemilihan smartphone berdasarkan
preferensi pengguna dengan mengintegrasikan metode K-Means sebagai
metode clustering dan TOPSIS sebagai metode perangkingan.

## 4. Metode

Data Mining:
- K-Means Clustering

Sistem Pendukung Keputusan:
- TOPSIS

## 5. Output

Sistem menghasilkan:
1. Kelompok smartphone berdasarkan karakteristiknya.
2. Nilai preferensi setiap smartphone.
3. Ranking smartphone.
4. Rekomendasi smartphone berdasarkan kebutuhan pengguna.
