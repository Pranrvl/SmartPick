```python
import pandas as pd
import numpy as np

print("Library berhasil diimport!")
```

```python
df = pd.read_csv("../data/processed/02_feature_engineering.csv")

print("Ukuran dataset:", df.shape)
display(df.head())
```

```python
df_kuesioner = pd.read_csv(
    "../data/raw/Kuesioner SPK Pemilihan Smartphone (Responses).csv"
)

print("Jumlah respons kuesioner:", len(df_kuesioner))
display(df_kuesioner.head())
```

```python
kolom_kriteria = {
    "price_idr": "Seberapa penting harga smartphone bagi Anda saat memilih smartphone?",
    "ram_gb": "Seberapa penting kapasitas RAM yang besar bagi Anda?",
    "storage_gb": "Seberapa penting kapasitas penyimpanan internal yang besar bagi Anda?",
    "clock_speed_ghz": "Seberapa penting performa prosesor bagi Anda?",
    "battery_mah": "Seberapa penting kapasitas baterai yang besar bagi Anda?",
    "rear_camera_main_mp": "Seberapa penting kualitas kamera belakang bagi Anda?",
    "charging_watt": "Seberapa penting fitur fast charging bagi Anda?"
}

df_preferensi = df_kuesioner[
    list(kolom_kriteria.values())
].copy()

df_preferensi.columns = list(kolom_kriteria.keys())

# Memastikan seluruh jawaban berupa angka
df_preferensi = df_preferensi.apply(
    pd.to_numeric, errors="coerce"
)

print("Jumlah respons yang digunakan:", len(df_preferensi))
display(df_preferensi.head())
```

```python
print("Jumlah nilai kosong:")
display(df_preferensi.isnull().sum())

if df_preferensi.isnull().any().any():
    raise ValueError(
        "Terdapat jawaban kosong pada pertanyaan kriteria. "
        "Periksa data kuesioner terlebih dahulu."
    )

if not df_preferensi.isin([1, 2, 3, 4, 5]).all().all():
    raise ValueError("Ditemukan nilai di luar skala 1 sampai 5.")

print("Data kuesioner valid!")
```

```python
rata_rata = df_preferensi.mean()

df_preferensi_mean = pd.DataFrame({
    "Kriteria": rata_rata.index,
    "Rata-rata Preferensi": rata_rata.values
})

print("Rata-rata preferensi pengguna:")
display(df_preferensi_mean.round(4))
```

```python
bobot = rata_rata / rata_rata.sum()

df_bobot = pd.DataFrame({
    "Kriteria": bobot.index,
    "Rata-rata": rata_rata.values,
    "Bobot": bobot.values
})

print("Bobot kriteria TOPSIS:")
display(df_bobot.round(4))

print("Total bobot:", round(bobot.sum(), 4))
```

```python
df_bobot.to_csv(
    "../outputs/bobot_kriteria_topsis.csv",
    index=False
)

print("Bobot kriteria berhasil disimpan!")
```

```python
kriteria = [
    "price_idr",
    "ram_gb",
    "storage_gb",
    "clock_speed_ghz",
    "battery_mah",
    "rear_camera_main_mp",
    "charging_watt"
]

jenis_kriteria = {
    "price_idr": "cost",
    "ram_gb": "benefit",
    "storage_gb": "benefit",
    "clock_speed_ghz": "benefit",
    "battery_mah": "benefit",
    "rear_camera_main_mp": "benefit",
    "charging_watt": "benefit"
}

# Memastikan seluruh kolom tersedia
kolom_hilang = [kolom for kolom in kriteria if kolom not in df.columns]

if kolom_hilang:
    raise ValueError(f"Kolom tidak ditemukan: {kolom_hilang}")

print("Kriteria TOPSIS:")
display(pd.DataFrame({
    "Kriteria": kriteria,
    "Jenis": [jenis_kriteria[k] for k in kriteria]
}))
```

```python
df_topsis = df.copy()

for kolom in kriteria:
    df_topsis[kolom] = pd.to_numeric(
        df_topsis[kolom], errors="coerce"
    )

jumlah_awal = len(df_topsis)

df_topsis = df_topsis.dropna(subset=kriteria).copy()

print("Jumlah alternatif awal:", jumlah_awal)
print("Jumlah alternatif valid:", len(df_topsis))
print("Jumlah alternatif yang dikeluarkan:", jumlah_awal - len(df_topsis))

display(df_topsis[kriteria].head())
```

```python
matriks_keputusan = df_topsis[kriteria].astype(float)

pembagi = np.sqrt(
    (matriks_keputusan ** 2).sum(axis=0)
)

if (pembagi == 0).any():
    raise ValueError("Terdapat kriteria dengan seluruh nilai nol.")

matriks_normalisasi = matriks_keputusan.div(pembagi, axis=1)

print("Matriks normalisasi TOPSIS:")
display(matriks_normalisasi.head())
```

```python
bobot_final = bobot.reindex(kriteria)

matriks_terbobot = matriks_normalisasi.mul(
    bobot_final, axis=1
)

print("Matriks terbobot berhasil dibuat!")
display(matriks_terbobot.head())
```

```python
solusi_ideal_positif = pd.Series(index=kriteria, dtype=float)
solusi_ideal_negatif = pd.Series(index=kriteria, dtype=float)

for kolom in kriteria:
    if jenis_kriteria[kolom] == "benefit":
        solusi_ideal_positif[kolom] = matriks_terbobot[kolom].max()
        solusi_ideal_negatif[kolom] = matriks_terbobot[kolom].min()
    else:
        solusi_ideal_positif[kolom] = matriks_terbobot[kolom].min()
        solusi_ideal_negatif[kolom] = matriks_terbobot[kolom].max()

print("Solusi ideal positif:")
display(solusi_ideal_positif.to_frame("A+"))

print("Solusi ideal negatif:")
display(solusi_ideal_negatif.to_frame("A-"))
```

```python
jarak_positif = np.sqrt(
    ((matriks_terbobot - solusi_ideal_positif) ** 2).sum(axis=1)
)

jarak_negatif = np.sqrt(
    ((matriks_terbobot - solusi_ideal_negatif) ** 2).sum(axis=1)
)

df_topsis["jarak_positif"] = jarak_positif
df_topsis["jarak_negatif"] = jarak_negatif

print("Perhitungan jarak solusi ideal:")
display(
    df_topsis[["jarak_positif", "jarak_negatif"]].head()
)
```

```python
penyebut = jarak_positif + jarak_negatif

if (penyebut == 0).any():
    raise ValueError("Tidak dapat menghitung nilai preferensi.")

df_topsis["nilai_preferensi"] = (
    jarak_negatif / penyebut
)

print("Nilai preferensi TOPSIS:")
display(df_topsis[["nilai_preferensi"]].head())
```

```python
df_topsis = df_topsis.sort_values(
    by="nilai_preferensi",
    ascending=False
).reset_index(drop=True)

df_topsis["ranking"] = range(1, len(df_topsis) + 1)

kolom_tampil = [
    kolom for kolom in [
        "smartphone_brand",
        "model",
        *kriteria,
        "nilai_preferensi",
        "ranking"
    ]
    if kolom in df_topsis.columns
]

print("10 Rekomendasi Smartphone Terbaik:")
display(df_topsis[kolom_tampil].head(10))
```

```python
df_topsis.to_csv(
    "../outputs/topsis_result_all.csv",
    index=False
)

print("Hasil TOPSIS berhasil disimpan!")
print("Lokasi: ../outputs/topsis_result_all.csv")
print("Jumlah alternatif:", len(df_topsis))
```

```python
# Membaca bobot kriteria TOPSIS
df_bobot = pd.read_csv("../outputs/bobot_kriteria_topsis.csv")

# Memeriksa struktur file
print("Struktur file bobot:")
print(df_bobot.columns.tolist())

# Menampilkan isi bobot
display(df_bobot)
```

```python

# Membaca dataset hasil clustering
df_cluster = pd.read_csv("../outputs/cluster_result.csv")

# Mengambil daftar kriteria dan bobot
kriteria = df_bobot["Kriteria"].tolist()
bobot = df_bobot.set_index("Kriteria")["Bobot"]

# Menentukan jenis kriteria
jenis_kriteria = {
    "price_idr": "cost",
    "ram_gb": "benefit",
    "storage_gb": "benefit",
    "clock_speed_ghz": "benefit",
    "battery_mah": "benefit",
    "rear_camera_main_mp": "benefit",
    "charging_watt": "benefit"
}

# Fungsi perhitungan TOPSIS
def hitung_topsis_cluster(group):

    group = group.copy()

    # Matriks keputusan
    X = group[kriteria].astype(float)

    # Normalisasi matriks
    pembagi = np.sqrt((X ** 2).sum(axis=0))
    R = X.div(pembagi.replace(0, np.nan), axis=1).fillna(0)

    # Normalisasi terbobot
    V = R.mul(bobot, axis=1)

    # Solusi ideal positif dan negatif
    ideal_positif = {}
    ideal_negatif = {}

    for kolom in kriteria:
        if jenis_kriteria[kolom] == "benefit":
            ideal_positif[kolom] = V[kolom].max()
            ideal_negatif[kolom] = V[kolom].min()
        else:
            ideal_positif[kolom] = V[kolom].min()
            ideal_negatif[kolom] = V[kolom].max()

    A_positif = pd.Series(ideal_positif)
    A_negatif = pd.Series(ideal_negatif)

    # Menghitung jarak ke solusi ideal
    D_positif = np.sqrt(((V - A_positif) ** 2).sum(axis=1))
    D_negatif = np.sqrt(((V - A_negatif) ** 2).sum(axis=1))

    # Menghitung nilai preferensi
    denominator = D_positif + D_negatif

    group["jarak_positif"] = D_positif
    group["jarak_negatif"] = D_negatif
    group["nilai_preferensi"] = np.divide(
        D_negatif,
        denominator,
        out=np.zeros(len(group)),
        where=denominator != 0
    )

    # Ranking dalam masing-masing cluster
    group["ranking_cluster"] = (
        group["nilai_preferensi"]
        .rank(method="first", ascending=False)
        .astype(int)
    )

    return group


print("Fungsi TOPSIS per cluster berhasil dibuat!")
```

```python
# Menghitung TOPSIS secara terpisah untuk setiap cluster
hasil_cluster = []

for cluster_id, group in df_cluster.groupby("cluster"):

    print(f"Menghitung TOPSIS Cluster {cluster_id}...")

    hasil = hitung_topsis_cluster(group)
    hasil_cluster.append(hasil)

# Menggabungkan hasil seluruh cluster
df_rekomendasi = pd.concat(
    hasil_cluster,
    ignore_index=True
)

print("\nPerhitungan TOPSIS per cluster berhasil!")
print("Total smartphone:", len(df_rekomendasi))

print("\nJumlah smartphone per cluster:")
display(df_rekomendasi["cluster"].value_counts().sort_index())

print("\n10 smartphone terbaik setiap cluster:")
display(
    df_rekomendasi
    .sort_values(["cluster", "ranking_cluster"])
    .groupby("cluster")
    .head(10)
)
```

```python
# Mengurutkan hasil berdasarkan cluster dan ranking
df_rekomendasi = df_rekomendasi.sort_values(
    by=["cluster", "ranking_cluster"]
)

# Menyimpan hasil rekomendasi
output_path = "../outputs/smartphone_recommendation.csv"

df_rekomendasi.to_csv(output_path, index=False)

print("Hasil rekomendasi berhasil disimpan!")
print("Lokasi file:", output_path)
print("Total smartphone:", len(df_rekomendasi))
```
